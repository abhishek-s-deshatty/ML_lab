```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram,linkage,fcluster
from sklearn.datasets import load_iris

df= load_iris()
x=df.data
y=df.target

scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

plt.figure(figsize=(13,6))
plt.title("agnes dendrogram")
dendrogram(linkage(x_scaled,method='ward'))
plt.xlabel('samples')
plt.ylabel('distance')
plt.show()

plt.figure(figsize=(13,6))
plt.title("diana dendrogram")
plt.xlabel("samples")
plt.ylabel("distance")
dendrogram(linkage(x_scaled,method='single'))
plt.show()

threshold = 2.5
agnes_cluster = fcluster(linkage(x_scaled,method='ward'),threshold,criterion='distance')
diana_cluster = fcluster(linkage(x_scaled,method='single'),threshold,criterion='distance')

plt.title("agnes cluster")
plt.scatter(x[:,0],x[:,1],c=agnes_cluster,cmap='viridis')
plt.xlabel("sepal_length")
plt.ylabel("sepal_width")
plt.show()

plt.title("diana cluster")
plt.scatter(x[:,0],x[:,1],c=diana_cluster,cmap='viridis')
plt.xlabel("sepal_length")
plt.ylabel("sepal_width")
plt.show()
```
